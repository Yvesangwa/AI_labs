### **Task 1: Create a DataFrame using Pandas**

### **Task 3: Handling Null Values**

```python
# Create a new DataFrame with some null values and fill them

# My Python code here:
new_data = {
    "Name" : ["John", "Eve", "Doe", "Mary"],
    "Age" : [29, None, 45, 39],
    "City" : ["New York", "San Francisco", None, "Chicago"]
}

new_df = pd.DataFrame(new_data)
print(new_df)
```

```python
# Fill in the missing value

new_df["Age"] = new_df["Age"].fillna(new_df["Age"].mean()).round().astype(int)
new_df["City"] = new_df["City"].fillna("Unknown")

print(new_df)
```

### **Task 5: Concatenation and Merging**

```python
# Concatenate two DataFrames vertically

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# My Python code here:

df_vertical_concat = pd.concat([df1, df2], ignore_index=True)

print(df_vertical_concat)
```

```python
# Merge the dataframe horizontally

df_horizontal_merge = pd.concat([df_vertical_concat, df3], axis = 1)

print(df_horizontal_merge)
```

### **Task 9: Iterators and Generators**

```python
# Create an iterator for the first 5 even numbers

# My Python code here:

# Create a list of the first five even numbers
even_numbers = [2,4,6,8,10]

# Iterate the list
iterate_numbers = iter(even_numbers)

print("Five even numbers:")
print(next(iterate_numbers))
print(next(iterate_numbers))
print(next(iterate_numbers))
print(next(iterate_numbers))
print(next(iterate_numbers))
      
```

```python
# Create a generator for the same

def generator_even():
    for number in range (1,6):
        yield number * 2

for value in generator_even():
    print(value)
```

### **Task 10: Map, Reduce, and Filter**

```python
# Use map to square all numbers in a list

numbers = [1, 2, 3, 4, 5]

# My python code here:
squared_numbers = list(map(lambda x: x ** 2, numbers))

print (squared_numbers)
```

```python
# Use reduce to find the product of all numbers in a list

numbers_product = reduce(lambda x, y: x*y, numbers)

print(numbers_product)
```

```python
# Use filter to get only even numbers from a list

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print (even_numbers)
```

### **Task 11: Object-Oriented Programming - Creating a Class**

```python
# Create a class called 'Rectangle' with attributes 'length' and 'width'


# Python code here:

# Create a Rectangle class

class Rectangle:
# Initialize the attributes (length and width)
    def __init__(self, length, width):
        self.length = length
        self.width = width

# Method to calculate area
    def area(self):
        return self.length * self.width
    
# Method to calculate perimeter
    def perimeter(self):
        return 2 * (self.length + self.width)
    
# Create two Rectangle objects
rectangle1 = Rectangle(15, 4)
rectangle2 = Rectangle(10, 5)

# Display the results

print("Rectangle 1 Area:", rectangle1.area())
print("Rectangle 1 Perimeter:", rectangle1.perimeter())

print("Rectangle 2 Area:", rectangle2.area())
print("Rectangle 2 Perimeter:", rectangle2.perimeter())
```

### **Task 12: Pandas Data Analysis**

```python
# Using the following DataFrame, perform these tasks:

df_employees = pd.DataFrame({
    'Name': ['John', 'Jane', 'Bob', 'Alice', 'Charlie'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [55000, 65000, 70000, 60000, 58000]
})

# 1. Find the average salary by department

average_salary_depart = df_employees.groupby("Department")["Salary"].mean().sort_values(ascending=False)

print(f"Average Salary by Department:\n{average_salary_depart}")
```

```python
# 2. Get the names of employees with salary > 60000

highly_paid_employees = df_employees[df_employees["Salary"] > 60000]["Name"]

print("Highly paid employees:\n", highly_paid_employees)
```

```python
# 3. Add a new column 'Bonus' which is 10% of salary

df_employees["Bonus"] = df_employees["Salary"] * 0.1

print(df_employees)
```

```python
# 4. Add a new column 'New_Salary' after adding bonus

df_employees["New_Salary"] = df_employees["Salary"] + df_employees["Bonus"]

print(df_employees)
```
