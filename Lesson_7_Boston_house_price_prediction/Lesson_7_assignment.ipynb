## **Supervised machine Learning - Regression**

### **_House Price Prediction Using Bolton Housing Data_**
This dataset provides information about house prices in Boston. House Price Prediction


In this project, I followed the basic machine learning process to build a regression model to predict house prices using the "Boston Housing Dataset" from sklearn.

The following is a description of each column in the dataset:

Dataset Features (Boston Housing):

* CRIM: Crime rate by town
* ZN: Proportion of residential land zoned for large lots
* INDUS: Proportion of non-retail business acres per town
* CHAS: Charles River dummy variable (1 if tract bounds river; 0 otherwise)
* NOX: Nitric oxide concentration (parts per 10 million)
* RM: Average number of rooms per dwelling
* AGE: Proportion of owner-occupied units built before 1940
* DIS: Weighted distances to five Boston employment centers
* RAD: Index of accessibility to radial highways
* TAX: Full-value property tax rate per `$10,000`
* PTRATIO: Pupil-teacher ratio by town
* B: Proportion of Black population
* LSTAT: Percentage of lower status of the population
* MEDV (Target): Median value of owner-occupied homes in `$1,000s`



**Dataset is from sklearn Datasets**

### **1. Importing libraries**

```python
# Imports libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

### **2. Data Collection and Loading**

```python
# Load the 'Boston Housing' dataset from sklearn and convert it into a pandas DataFrame.

# Load dataset and convert to DataFrame:

from sklearn.datasets import fetch_openml
boston_data = fetch_openml(name='boston', version=1, as_frame=True, return_X_y=False, parser='auto')
boston_df = boston_data.frame
```

### **3. Quick Check of Data**

```python
# Display the first few rows of the dataset to understand its structure.

boston_df.head()
```

```python
# Check the features and target variable. Inspect data types 

boston_df.info()
```

```python
# Display the statistical properties

boston_df.describe()
```

```python
# Change data type for categorical variables.

boston_df["CHAS"] = boston_df["CHAS"].astype(float)
boston_df["RAD"] = boston_df["RAD"].astype(float)
```

### **4. EDA and Data Preprocessing**

```python
# Check for missing/null values

print(boston_df.isnull().sum())
```

```python
# Visualize the data. Create scatter plots to see the relationship between independent features and the target variable.
# Example: Use `plt.scatter()` to visualize the relationship between features like 'RM' (average number of rooms) and the target ('MEDV').

fig = px.scatter(
    boston_df, 
    x="RM",
    y="MEDV",
    title="Average number of rooms and Median value of owner-occupied homes")

fig.show()
```

```python
# Create a function to automate scatter plots for multiple features vs the target variable.

# Define the function

def visualizeFeatureTarget(boston_df, x):
    fig = px.scatter(
        boston_df,
        x = x,
        y = "MEDV",
        trendline = "ols")
    fig.show()

# Call the function

visualizeFeatureTarget(boston_df, "RM")
visualizeFeatureTarget(boston_df, "LSTAT")
visualizeFeatureTarget(boston_df, "AGE")
visualizeFeatureTarget(boston_df, "CRIM")

    
```

## **5. ML Model Training**

```python
# Split the dataset into training and testing sets

# Define X (features) and y (target) and perform the train-test split:

x = boston_df.drop(columns=["MEDV", "B"])
y = boston_df["MEDV"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
# Create the regression models:

lr = LinearRegression()
dt = DecisionTreeRegressor(random_state=42)
rf = RandomForestRegressor(n_estimators=100, random_state=42)
```

```python
# Train the models on the training data

lr.fit(x_train, y_train)
dt.fit(x_train, y_train)
rf.fit(x_train, y_train)
```

### **6. Model Evaluation**

```python
# Predict on the test set
y_pred_lr = lr.predict(x_test)
y_pred_dt = dt.predict(x_test)
y_pred_rf = rf.predict(x_test)
```

```python
# Evaluate Linear Regression
mse_lr = mean_squared_error(y_test, y_pred_lr)
rmse_lr = np.sqrt(mse_lr)
r2_lr = r2_score(y_test, y_pred_lr)
```

```python
# Evaluate Decision Tree
mse_dt = mean_squared_error(y_test, y_pred_dt)
rmse_dt = np.sqrt(mse_dt)
r2_dt = r2_score(y_test, y_pred_dt)
```

```python
# Evaluate Random Forest
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mse_rf)
r2_rf = r2_score(y_test, y_pred_rf)
```

```python
print("Linear Regression")
print("MSE:", mse_lr)
print("RMSE:", rmse_lr)
print("R²:", r2_lr)

print("\nDecision Tree")
print("MSE:", mse_dt)
print("RMSE:", rmse_dt)
print("R²:", r2_dt)

print("\nRandom Forest")
print("MSE:", mse_rf)
print("RMSE:", rmse_rf)
print("R²:", r2_rf)
```
