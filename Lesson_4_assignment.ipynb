### **Task 1: Create a DataFrame using Pandas**

```python
# Python and Pandas Assessment Notebook

import pandas as pd
import numpy as np
from functools import reduce
```

```python
# Create a DataFrame with the following data:
# Name: Alice, Bob, Charlie, David
# Age: 25, 30, 35, 28
# City: New York, San Francisco, Los Angeles, Chicago

# Python codes here:
data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [25, 30, 35, 28],
    "City": ["New York", "San Francisco", "Los Angeles", "Chicago"]}

# Convert to dataframe

df = pd.DataFrame(data)

# Print the dataframe
print(df)
```

### **Task 2: Row and Column Manipulation**

```python
# Drop the 'City' column from the DataFrame created in Task 1

# My Python code here:

df = df.drop(["City"], axis=1)

print(df)
```

### **Task 3: Handling Null Values**

```python
# Create a new DataFrame with some null values and fill them

# My Python code here:
new_data = {
    "Name" : ["John", "Eve", "Doe", "Mary"],
    "Age" : [29, None, 45, 39],
    "City" : ["New York", "San Francisco", None, "Chicago"]
}

new_df = pd.DataFrame(new_data)
print(new_df)
```

```python
# Fill in the missing value

new_df["Age"] = new_df["Age"].fillna(new_df["Age"].mean()).round().astype(int)
new_df["City"] = new_df["City"].fillna("Unknown")

print(new_df)
```

### **Task 4: GroupBy and Describe**

```python
# Using the following DataFrame, group by 'Category' and describe the 'Value' column

df_group = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'C'],
    'Value': [10, 20, 15, 25, 30, 35]
})

# My Python code here:

df_category = df_group.groupby("Category")["Value"].describe()

print(df_category)
```

### **Task 5: Concatenation and Merging**

```python
# Concatenate two DataFrames vertically

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# My Python code here:

df_vertical_concat = pd.concat([df1, df2], ignore_index=True)

print(df_vertical_concat)
```

```python
# Merge the dataframe horizontally

df_horizontal_merge = pd.concat([df_vertical_concat, df3], axis = 1)

print(df_horizontal_merge)
```
